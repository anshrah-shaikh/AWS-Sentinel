```python
#importing libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.ensemble import IsolationForest
from sklearn.preprocessing import StandardScaler
import joblib
```

```python
import requests
import time
```

```python
cities = {
    "Mumbai": (19.0760, 72.8777),
    "Delhi": (28.6139, 77.2090),
    "Chennai": (13.0827, 80.2707),
    "Kolkata": (22.5726, 88.3639),
    "Bengaluru": (12.9716, 77.5946)
}

start_date = "2021-01-01"
end_date = "2025-12-31"
```

```python
all_data = []

for city, (latitude, longitude) in cities.items():

    print(f"Downloading data for {city}...")

    url = "https://archive-api.open-meteo.com/v1/archive"

    params = {
        "latitude": latitude,
        "longitude": longitude,
        "start_date": start_date,
        "end_date": end_date,
        "hourly": (
            "temperature_2m,"
            "relative_humidity_2m,"
            "precipitation,"
            "surface_pressure,"
            "wind_speed_10m,"
            "wind_direction_10m"
        ),
        "timezone": "Asia/Kolkata"
    }

    response = requests.get(url, params=params)

    if response.status_code == 200:

        data = response.json()

        df = pd.DataFrame(data["hourly"])

        df["city"] = city

        all_data.append(df)

        print(f"✓ {city} downloaded successfully!")

    else:
        print(f"✗ Failed to download {city}")

    time.sleep(1)
```

```python
weather_df = pd.concat(all_data, ignore_index=True)

print("Dataset shape:", weather_df.shape)

weather_df.head()
```

```python
import numpy as np
import pandas as pd
from sklearn.metrics import precision_score, recall_score, f1_score

# Actual anomaly labels
y_true = final_test_df["injected_anomaly"].values

# Thresholds to test
thresholds = np.linspace(
    anomaly_scores.min(),
    anomaly_scores.max(),
    100
)

results = []

for threshold in thresholds:

    # Lower score than threshold = anomaly
    y_pred = (anomaly_scores < threshold).astype(int)

    precision = precision_score(
        y_true,
        y_pred,
        zero_division=0
    )

    recall = recall_score(
        y_true,
        y_pred,
        zero_division=0
    )

    f1 = f1_score(
        y_true,
        y_pred,
        zero_division=0
    )

    results.append({
        "threshold": threshold,
        "precision": precision,
        "recall": recall,
        "f1_score": f1
    })

results_df = pd.DataFrame(results)

print("Top 10 thresholds by F1 score:\n")

print(
    results_df
    .sort_values("f1_score", ascending=False)
    .head(10)
)
```

```python
best_row = results_df.loc[
    results_df["f1_score"].idxmax()
]

best_threshold = best_row["threshold"]

print("🏆 BEST THRESHOLD:", best_threshold)

print("\nPrecision:", best_row["precision"])
print("Recall:", best_row["recall"])
print("F1 Score:", best_row["f1_score"])
```

```python
# Create final predictions using the best threshold

final_predictions = (
    anomaly_scores < best_threshold
).astype(int)

# Save predictions
final_test_df["optimized_model_anomaly"] = final_predictions

print(
    final_test_df[
        "optimized_model_anomaly"
    ].value_counts()
)
```

```python
import numpy as np
import pandas as pd

def predict_weather_anomaly(
    temperature,
    humidity,
    precipitation,
    pressure,
    wind_speed,
    wind_direction,
    temperature_change=0,
    humidity_change=0,
    precipitation_change=0,
    pressure_change=0,
    wind_speed_change=0,
    wind_direction_change=0
):

    # Create one input row using the same features
    input_data = pd.DataFrame([{
        "temperature_2m": temperature,
        "relative_humidity_2m": humidity,
        "precipitation": precipitation,
        "surface_pressure": pressure,
        "wind_speed_10m": wind_speed,
        "wind_direction_10m": wind_direction,

        "temperature_2m_change": temperature_change,
        "relative_humidity_2m_change": humidity_change,
        "precipitation_change": precipitation_change,
        "surface_pressure_change": pressure_change,
        "wind_speed_10m_change": wind_speed_change,
        "wind_direction_10m_change": wind_direction_change
    }])

    # Make sure feature order matches training
    input_data = input_data[improved_features]

    # Scale the input
    scaled_input = final_scaler.transform(input_data)

    # Get anomaly score
    anomaly_score = final_model.decision_function(
        scaled_input
    )[0]

    # Apply our optimized threshold
    if anomaly_score < best_threshold:
        status = "ANOMALY"
    else:
        status = "NORMAL"

    return {
        "status": status,
        "anomaly_score": anomaly_score,
        "threshold_used": best_threshold
    }


print("Prediction function created successfully! ✅")
```

```python
def get_anomaly_severity(anomaly_score, threshold):

    # Calculate how far below the threshold the reading is
    deviation = threshold - anomaly_score

    if anomaly_score >= threshold:
        return "NORMAL"

    elif deviation < 0.05:
        return "LOW"

    elif deviation < 0.15:
        return "MEDIUM"

    else:
        return "HIGH"
```

```python
def predict_weather_anomaly_detailed(
    temperature,
    humidity,
    precipitation,
    pressure,
    wind_speed,
    wind_direction,
    temperature_change=0,
    humidity_change=0,
    precipitation_change=0,
    pressure_change=0,
    wind_speed_change=0,
    wind_direction_change=0
):

    # Get model prediction
    result = predict_weather_anomaly(
        temperature,
        humidity,
        precipitation,
        pressure,
        wind_speed,
        wind_direction,
        temperature_change,
        humidity_change,
        precipitation_change,
        pressure_change,
        wind_speed_change,
        wind_direction_change
    )

    severity = get_anomaly_severity(
        result["anomaly_score"],
        result["threshold_used"]
    )

    result["severity"] = severity

    result["weather_reading"] = {
        "temperature": temperature,
        "humidity": humidity,
        "precipitation": precipitation,
        "pressure": pressure,
        "wind_speed": wind_speed,
        "wind_direction": wind_direction
    }

    return result


print("Detailed prediction function created! ✅")
```
